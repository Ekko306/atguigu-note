# 默认配置

1.2版本新加的内容，这里只讲解，不演示

逻辑简单，有的时候，多个节点具备相同的执行规则比如 a b c三个节点都需要`retry_policy=RetryPolicy(max_attempts=3)`填写参数三次才能生效

但是1.2版本加了新功能， set_node_defaults 为所有节点设置默认配置，全图默认配置，
- retry_policy 重试
- timeout 超时
- error_handler 错误处理
- cache_policy 缓存

只需要设置一遍，后续其他节点没有额外声明，都是一样的

https://docs.langchain.com/oss/python/langgraph/fault-tolerance#graph-defaults

也是可选项，是1.2版本新特性，可以看官方文档
