# Langgraph中的边
学习Langgraph中边的总结，前面有很多边的用法，这里系统总结

边： 定义节点之间的依赖关系，当着节点执行完了，下一个节点执行谁
1. 普通边， a执行完，执行b
2. 条件边，一般添加路由函数，路由函数前面有很多，表示这个节点执行完后，进入路由函数，路由函数的返回值，决定下一个节点是谁，一般根据状态里面的值进行条件判断，路由函数里的返回值，通常指代节点的名称
3. 特别的：入口点START，停止点END， 启动点一定要有START，不然不知道先运行谁，然后执行node_a节点， 
    1. 也可以使用set_entry_point指定起订，  但是推荐使用add_edge, 不涉及其他api
4. 特别的条件入口点，就是START+add_conditional_edges，混合而已

注意点：
1. A下面有bcd，在超步里面是同一个编号，并行执行



# 总结条件边方式  

在实际开发中，建议同一个节点尽量只选择一种路由机制：
   - 固定流转
        - 普通边
   - 动态流转
        - 条件边 add_conditional_edges, 定义router函数
            - 要么固定返回'node_a'
            - 要么用list.push(Send(...))
        - 自定义router_node， 使用`Command(goto)` （倒反天罡，在node逻辑里就跳转了）

    三选一


之前笔记：
两个典型用法
1. send 根据当前状态可以触发哪些任务 （用的多一点，因为不太方便用静态方式实现，专门属于动态分支）
2. command 跳转到下游节点（用静态分支也可以实现）
