# 拓展内容AgentState
拓展内容和LangChain使用相关

也可以使用Agent作为状态， 从langchian.agents.middleware,types.AgentState读取

结论：不推荐这样使用，  agent源代码下面

```
class AgentState(TypedDict, Generic[ResponseT]):
    """State schema for the agent."""

    messages: Required[Annotated[list[AnyMessage], add_messages]]
    jump_to: NotRequired[Annotated[JumpTo | None, EphemeralValue, PrivateStateAttr]]
    structured_response: NotRequired[Annotated[ResponseT, OmitFromInput]]
```

三个参参数

1. messages  和MessagesState一样，
2. jump_to 跳转字段，主要用于在Agent中间件跳转到那个位置，但在langgraph不通用
   1langgraph里用的是command(goto="node_name") 不一样，会导致，使用agent state会有误会，谢了jump_to，结果没效果，不利于框架
    1.
3. structured response： 最终生成结构化输出，也是专门为langchain使用的，  不能直接langgraph使用， 不太适配
（可以用，但是使用要用好，不能直接导入，但要用好，不能只导入，还要配套的东西拿过来，这样比较麻烦）

选择langgraph更底层， 不要使用langchain的，否则直接用langchain，所以推荐MessagesAgent
