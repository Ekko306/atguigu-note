```python
from langgraph.graph import StateGraph,START,END
from typing import TypedDict, Annotated
from operator import add


# 1 定义状态 state ，要传入父类， 推荐用TypeDict，可以使用字典，  定义好STATE，表示类被当成字典使用
class OverAllState(TypedDict): # 字典就是python自带的字典，官方推荐用字典作为基础的类型

    # Annotated 有注解的， 在编译器没有用，在langgrah有用
    # 日志类型还是 list[str]  更新的方式不是覆盖 add 是追加。 原来的类型：用logs: list[str]就行，但是在langraph不能更新
    logs: Annotated[list[str],add]  # 更新的时候 添加， 添加更新， add要从operator包导入， 追加的集合
    cur_id: str # 当前运行的流程位置
    # cur_id: Annotated[str, add] # 版本二



# 2 定义节点，定义执行的函数
def node_1(state:OverAllState) -> OverAllState:
    pre_id = state["cur_id"] # 获取当前的id
    return {  # 返回结果也要一个字典类型
        "logs": ["node_1 运行完毕"],
        "cur_id": pre_id + ", node_1", # 返回的时候拼接，
        # "cur_id": ", node_1" # 版本二

    }

def node_2(state:OverAllState) -> OverAllState:
    pre_id = state["cur_id"]
    return {
        "logs": ["node_2 运行完毕"],
        "cur_id": pre_id + ", node_2",
        # "cur_id": ", node_1" # 版本二

    }

# 3 定义边     （边是依赖于图的构成再来执行的，必须要先有一个主体，要导入 StateGraph,START,END）
# 3.1 创建图 获取建造者
builder = StateGraph( state_schema=OverAllState)  # 先创建graph
# 3.2 添加节点
builder.add_node(node_1)  # 添加node
builder.add_node(node_2)
# 3.3 添加边
builder.add_edge(START,"node_1")  # 添加edge
builder.add_edge("node_1","node_2")
builder.add_edge("node_2",END)

# 4 获取图
graph = builder.compile()

# 5 运行图
result = graph.invoke({"cur_id":"start"}) # 要传入初始的状态
print(result)

# cur_id会被覆盖掉，所以要手动拼接
# 但是logs是add类型，会把返回的值做一个处理
```

```python
raw_mermaid = graph.get_graph().draw_mermaid()
print(raw_mermaid)
```

```python
png_bytes = graph.get_graph().draw_mermaid_png() # 调用base_url的地址api，把上面的逻辑代码转换成二进制数据，然后自己处理
# print(png_bytes)


# 方法1 转成base64
import base64
b64 = base64.b64encode(png_bytes).decode()
data_uri = f"data:image/png;base64,{b64}"
print(data_uri)  # 复制到浏览器地址栏即可查看
```

```python
# 方法2 jupyter显示
from IPython.display import display, Image
png = Image(png_bytes)
display(png)
```

```python
# 方法3 本地保存
png_filename = '../first_demo_graph_1.png'
with open(png_filename, "wb") as f:
    f.write(png_bytes)
```
