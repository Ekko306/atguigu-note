# 3.2 State Reducer
状态合并更新机制， 前面结构比较简单，不能实现归一合并的重要性

不只简单的串联，还要解决多个复杂的结构合并到一个节点，谁覆盖谁，要有稳定的核心的机制

reducer解决这个问题

- 函数签名：  (left, right)
- 注解定义
- 默认行为

```python
# 定义reducer

# left： 从最开始的位置合并到当前节点的位置的值
# right: 当前节点的值
# 返回值：合并后的值
def my_reducer(left: list[str], right: list[str]) -> list[str]:
    return left + right # python里面可以直接list集合相加


# 1. node1 运行之后的值
left = ["start", "node_1 运行完毕"]
right = ["node_2 运行完毕"]

# 2. 合并
merged = my_reducer(left, right)
print(merged)

# 串联不明显， 如果1 2 3 同时指向 node， 一定要有核心的reducer， 不然覆盖数据了
```
